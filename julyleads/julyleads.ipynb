```python
import pandas as pd
df = pd.read_excel("july leads.xlsx")
```

```python
import re
from collections import Counter

not_converted = df[df["Convert or not converted"] == "Not Converted"]

# Take Excel columns G and H using position (not name)
text = " ".join(
    not_converted.iloc[:,6].astype(str) + " " +
    not_converted.iloc[:,7].astype(str)
)

# Find words
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

# Count common words
common_words = Counter(words)

print(common_words.most_common(20))
```
