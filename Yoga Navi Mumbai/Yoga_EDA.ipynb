```python
import pandas as pd

df = pd.read_csv("Yoga Leads.csv")
df.head()
```

# Cleaning raw dataset

```python
df["Contact"] = df["Contact"].str.replace("·", "", regex=False).str.strip()
df["Industry"] = df["Industry"].str.replace("·", "", regex=False).str.strip()
```

```python
import numpy as np

df["Contact"] = df["Contact"].replace(["", " ", "nan"], np.nan)
df["Website"] = df["Website"].replace(["", " ", "nan"], np.nan)

df = df[df["Contact"].notna() | df["Website"].notna()]
```
