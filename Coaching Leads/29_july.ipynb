```python
from google.colab import files
import pandas as pd

uploaded = files.upload()

df = pd.read_excel("Teaching Institute.xlsx")

# View the first few rows
df.head()
```

```python
# Remove the middle dot and clean whitespace
df["address"] = df["address"].fillna("").str.replace("·", "", regex=False).str.strip()
df["address 2"] = df["address 2"].fillna("").str.replace("·", "", regex=False).str.strip()

# Create full address
df["Full Address"] = (
    df["address"] + ", " + df["address 2"]
).str.strip(", ")
```

```python
df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Count industries
industry_counts = df["industry"].fillna("Unknown").value_counts().reset_index()
industry_counts.columns = ["Industry", "Count"]

# Plot
plt.figure(figsize=(12, 8))
sns.barplot(data=industry_counts, x="Count", y="Industry")

plt.title("Number of Businesses by Industry")
plt.xlabel("Count")
plt.ylabel("Industry")

plt.tight_layout()
plt.show()
```
